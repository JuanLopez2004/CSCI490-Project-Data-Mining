# How to Work with Jupyter Notebooks

Welcome to your first Jupyter notebook! This tutorial will guide you through the basics of creating and working with Jupyter notebooks.

## What is a Jupyter Notebook?
A Jupyter notebook is an interactive document that combines code, text, and visualizations in a single file. It's perfect for data analysis, prototyping, and educational purposes.

## 1. Install Jupyter Notebook

Jupyter Notebook can be installed using pip or conda package managers:

```bash
# Using pip
pip install jupyter

# Using conda
conda install jupyter

# Or install with additional packages
pip install jupyter pandas matplotlib numpy seaborn
```

✅ **Good news**: We've already installed Jupyter in your environment!

## 2. Launch Jupyter Notebook

There are several ways to launch Jupyter Notebook:

### Method 1: Command Line
```bash
jupyter notebook
```

### Method 2: VS Code (Recommended)
- Open a `.ipynb` file in VS Code
- VS Code will automatically start the Jupyter kernel
- Work directly in the editor with syntax highlighting and IntelliSense

### Method 3: Browser Interface
- Run `jupyter notebook` in terminal
- Copy the URL with token to your browser
- Navigate through the file browser interface

## 3. Create a New Notebook

### In VS Code:
1. Right-click in the file explorer
2. Select "New File"
3. Name it with `.ipynb` extension (e.g., `my_notebook.ipynb`)
4. VS Code will open it as a Jupyter notebook

### In Browser:
1. Click "New" button
2. Select "Python 3" (or your preferred kernel)
3. A new notebook will open

### Key Interface Components:
- **Cells**: Containers for code or markdown
- **Kernel**: The Python process running your code
- **Toolbar**: Contains run, save, and other actions

## 4. Add and Execute Code Cells

Let's create some code cells and execute them. Try running the cells below!

````python
# Basic Python operations
print("Hello, Jupyter!")
print("This is your first code cell")

# Variables and calculations
x = 10
y = 20
result = x + y
print(f"The sum of {x} and {y} is {result}")
````

````python
# Import libraries (we installed these earlier)
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Create some sample data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'Score': [85, 92, 78, 96]
}

df = pd.DataFrame(data)
print("Sample DataFrame:")
print(df)
````

````python
# Create a simple visualization
plt.figure(figsize=(8, 5))
plt.bar(df['Name'], df['Score'], color=['blue', 'green', 'red', 'purple'])
plt.title('Sample Scores by Person')
plt.xlabel('Name')
plt.ylabel('Score')
plt.ylim(0, 100)

# Display the plot
plt.show()

print("🎉 Congratulations! You've created your first data visualization in Jupyter!")
````

## 5. Add Markdown Cells

Markdown cells are perfect for adding documentation, explanations, and formatted text to your notebooks.

### Markdown Features:
- **Bold text** and *italic text*
- `Code snippets`
- Lists (like this one!)
- [Links](https://jupyter.org/)
- Mathematical equations: $E = mc^2$

### Tips for Code Execution:
- **Shift + Enter**: Run current cell and move to next
- **Ctrl + Enter**: Run current cell and stay
- **Alt + Enter**: Run current cell and insert new cell below

### Cell Types:
- **Code**: Python code that gets executed
- **Markdown**: Text, documentation, and formatting
- **Raw**: Plain text (not executed or formatted)

## 6. Save and Export Notebooks

### Saving Your Work:
- **Auto-save**: Notebooks auto-save periodically
- **Manual save**: Ctrl+S (VS Code) or File > Save
- **Checkpoint**: Creates a backup copy

### Export Options:
Your notebooks can be exported to various formats:

1. **HTML**: For sharing and viewing in browsers
2. **PDF**: For formal documents and reports
3. **Python script**: Extract just the code
4. **Slides**: For presentations

### Working with Your CSV File:
Now that you know the basics, you can load and analyze your `original.csv` file!

````python
# Let's try loading your CSV file!
# Make sure the file exists first
import os

# Check if the CSV file exists
csv_file = "original.csv"
if os.path.exists(csv_file):
    print(f"✅ Found {csv_file}")
    
    # Load the CSV file
    try:
        data = pd.read_csv(csv_file)
        print(f"\n📊 Data shape: {data.shape}")
        print(f"📋 Columns: {list(data.columns)}")
        print("\n🔍 First few rows:")
        print(data.head())
    except Exception as e:
        print(f"❌ Error loading CSV: {e}")
else:
    print(f"❌ Could not find {csv_file}")
    print("📁 Current directory contents:")
    print(os.listdir("."))
````
