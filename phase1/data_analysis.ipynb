# Heart Failure Data Analysis

Analysis of heart failure clinical records dataset for prediction modeling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load data
data = pd.read_csv('original.csv')

print("Dataset shape:", data.shape)
print("\nFirst 5 rows:")
data.head()
```

```python
# Basic statistics
print("Dataset info:")
print(data.info())
print("\nSummary statistics:")
print(data.describe())
```

```python
# Target variable distribution
plt.figure(figsize=(8, 5))
data['DEATH_EVENT'].value_counts().plot(kind='bar')
plt.title('Death Event Distribution')
plt.xlabel('Death Event (0=Survived, 1=Died)')
plt.ylabel('Count')
plt.show()

print("Death event counts:")
print(data['DEATH_EVENT'].value_counts())
```

```python
# Age distribution by outcome
plt.figure(figsize=(10, 6))
plt.subplot(1, 2, 1)
plt.hist(data[data['DEATH_EVENT']==0]['age'], alpha=0.7, label='Survived', bins=15)
plt.hist(data[data['DEATH_EVENT']==1]['age'], alpha=0.7, label='Died', bins=15)
plt.title('Age Distribution by Outcome')
plt.xlabel('Age')
plt.ylabel('Count')
plt.legend()

plt.subplot(1, 2, 2)
plt.boxplot([data[data['DEATH_EVENT']==0]['age'], data[data['DEATH_EVENT']==1]['age']], 
           labels=['Survived', 'Died'])
plt.title('Age by Outcome')
plt.ylabel('Age')

plt.tight_layout()
plt.show()
```

```python
# Correlation matrix
plt.figure(figsize=(12, 8))
correlation_matrix = data.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Feature Correlation Matrix')
plt.show()
```
